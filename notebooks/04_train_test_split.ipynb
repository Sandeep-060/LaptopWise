```python
import pandas as pd
df = pd.read_csv("../data/processed/laptop_selected.csv")
df
```

```python
print(df.shape)
print(df.columns.tolist())
print("Target missing values:", df["Price (Rs)"].isna().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
feature_columns = [
    "Brand",
    "Series",
    "Thickness",
    "Weight",
    "Operating System",
    "Display Size",
    "Display Touchscreen",
    "Processor",
    "Graphic Processor",
    "RAM_Capacity_GB",
    "RAM Type",
    "SSD Capacity",
    "HDD Capacity",
    "Battery Capacity",
    "Fingerprint scanner"
]

target_column = "Price (Rs)"
```

```python
print("Predictor count:", len(feature_columns))
print("Target:", target_column)
```

```python
X = df[feature_columns]
y = df[target_column]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
print("Training rows:", len(X_train))
print("Testing rows :", len(X_test))
print("Total rows   :", len(X_train) + len(X_test))
print("Original rows:", len(X))
```

```python
print("Index overlap:", len(set(X_train.index) & set(X_test.index)))
```

```python
print("Train percentage:", len(X_train) / len(X) * 100)
print("Test percentage :", len(X_test) / len(X) * 100)
```
