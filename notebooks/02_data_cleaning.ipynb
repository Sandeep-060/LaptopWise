# Invalid & Inconsistent Data

```python
categorical_columns = df.select_dtypes(include=["object", "string"]).columns

print("Categorical columns:", len(categorical_columns))

for column in categorical_columns:
    print(column)
```

```python
for column in categorical_columns:
    values = df[column].dropna().astype(str)
    original_unique = values.nunique()
    print(column,":",original_unique)
```

```python
col=[]
org=[]
nor=[]
for column in categorical_columns:
    col.append(column)
    values = df[column].dropna().astype(str)
    original_unique = values.nunique()
    normalized_unique = (values.str.strip().str.lower().nunique())
    org.append(original_unique)
    nor.append(normalized_unique)

pd.DataFrame({"column":col,"unique_values":org,"normalized_values":nor})
```

```python
inconsistent_columns = []

for column in categorical_columns:

    values = df[column].dropna().astype(str)
    original_unique = values.nunique()
    normalized_unique = (
        values
        .str.strip()
        .str.lower()
        .nunique()
    )

    if normalized_unique < original_unique:
        inconsistent_columns.append(column)

print("Potentially inconsistent columns:")
print(inconsistent_columns)
print("total:",len(inconsistent_columns))
```

```python
for column in inconsistent_columns:

    values = df[column].dropna().astype(str)

    normalized = values.str.strip().str.lower()

    comparison = pd.DataFrame({
        "Original": values,
        "Normalized": normalized
    })

    conflicts = comparison[
        comparison["Original"] != comparison["Normalized"]
    ].drop_duplicates()

    print(f"\n{'=' * 70}")
    print(column)
    print(f"{'=' * 70}")

    print(conflicts.head(20))
```

```python
for column in inconsistent_columns:
    df[column] = (
        df[column]
        .astype("string")
        .str.strip()
        .str.lower()
    )
```

```python
for column in inconsistent_columns:

    values = df[column].dropna().astype(str)

    normalized = values.str.strip().str.lower()

    if values.nunique() != normalized.nunique():
        print(column)
        print("Still has formatting differences")
```

# Text-Encoded Numerical Specifications

```python
text_columns = df.select_dtypes(
    include=["object", "string"]
).columns

print("Number of text columns:", len(text_columns))

for column in text_columns:
    print(f"\n--- {column} ---")
    print(df[column].dropna().head(10).to_list())
```

```python
# Direct Numeric Extraction

measurement_columns = [
    "Thickness",
    "Weight",
    "Display Size",
    "Pixel Density",
    "Clock-speed",
    "Cache",
    "RAM Speed",
    "SSD Capacity",
    "Battery Cell",
    "Power Supply",
    "Refresh Rate",
    "Brightness",
    "Graphics Memory",
    "Expandable Memory",
    "HDD Capacity",
    "HDD Speed(RPM)",
    "Battery Life",
    "Battery Capacity"
]
```

```python
df[measurement_columns].head(10)
```

```python
for column in measurement_columns:
    df[column] = (
        df[column]
        .astype("string")
        .str.extract(r"([-+]?\d*\.?\d+)", expand=False)
        .astype(float)
    )
```
